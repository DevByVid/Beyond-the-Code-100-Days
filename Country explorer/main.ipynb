```python
import requests
```

```python
url = "https://api.restcountries.com/countries/v5/names.common/India"

headers = {
    "Authorization": "Bearer rc_live_demo"
}

response = requests.get(url, headers=headers)
print(response.status_code)
```

```python
data=response.json()
print(data)
```

```python
country = data['data']['objects'][0]

print(country['names']['common'])
```

```python
country=data['data']['objects'][0]
print("Country:", country['names']['common'])
print("Capital:", country['capitals'][0]['name'])
print("Region:", country['region'])
print("Population:", country['population'])
             
```

```python

import requests

country_name = input("Enter a country name: ")

url = f"https://api.restcountries.com/countries/v5/names.common/{country_name}"

try:
    response = requests.get(url, headers=headers, timeout=10)
    response.raise_for_status()

    result = response.json()

    if '_demo' in result['data']:
        print("\n⚠️ Demo API response received.")
        print("The API returned sample data, not your requested country.")

    else:
        countries = result['data']['objects']

        if countries:
            country = countries[0]
            print("\n🌍 Country:", country['names']['common'])
        else:
            print("Country not found!")

except requests.exceptions.RequestException as e:
    print("\nNetwork or API error:", e)

except (ValueError, KeyError, IndexError, TypeError) as e:
    print("\nData processing error:", e)

finally:
    print("\n✨ Country Explorer session completed!")
```
